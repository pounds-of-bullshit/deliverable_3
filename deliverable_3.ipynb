## Task 1: Character class

The `Character` class represents a character equipped with a name, hit points (HP), base damage and an optional weapon. The `take_damage()` and `heal()` methods control changes to HP, `attack()` deals damage, and `introduce()` introduces each character. The read-only properties `hp` and `is_alive` are used to inspect each character's state.

- **Attributes:** `name`, `_hp`, `hp_max`, `damage`, `weapon`

- **Methods:** `take_damage()`, `heal()`, `attack()`, `introduce()`, `__repr__()`, `__str__()`

- **Properties:** `hp`, `is_alive`

```python
# A demonstration of the Character class

# Creates an example character without a weapon
ex = Character(name="Aria", hp_max=100, damage=10)
# Prints the character introduction
ex.introduce()

# Displays the character, the starting HP and whether or not they're alive
print(ex)
print(f"Initial HP: {ex.hp}")
print(f"Alive: {ex.is_alive}")

# Tests damage and healing
ex.take_damage(30)
print(f"HP after damage: {ex.hp}")

ex.heal(15)
print(f"HP after healing: {ex.hp}")

# Tests HP boundaries
ex.heal(200)
print(f"HP after excessive healing: {ex.hp}")

ex.take_damage(200)
print(f"HP after excessive damage: {ex.hp}")
print(f"Alive: {ex.is_alive}")
```

The results show that the character Aria starts with 100 HP. Taking damage reduces it, while healing increases it. The HP could also not go above the starting HP, neither below 0. When Aria dies (HP = 0), the `is_alive` property returns `False`.


The special methods `__repr__()` and `__str__()` determine how the character is displayed in Python.

```python
# A demonstration of the special methods

print(repr(ex))
print(str(ex))
```

## Task 2: Dice class

The `Dice` class creates any n-sided dice, where the default number of sides is 6. The `roll()` method returns a random integer between 1 and the n number of sides, while the `__call__()` method allows a dice object to be used like a function.

- **Attributes:** `sides`

- **Methods:** `roll()`, `__repr__()`, `__call__()`

```python
# A demonstration of the Dice class

# Creates a dice with n numbers of sides
d6 = Dice() # 6 sides
d20 = Dice(sides=20) # 20 sides

print(repr(d6))
print(repr(d20))
print(f"Roll using roll(): {d6.roll()}")
print(f"Roll using __call__(): {d20()}")

# Repeated dice rolls
# Sets a seed to repeat the same random results
random.seed(42)

# Generates 10 dice rolls for the six-sided one
rolls = [d6.roll() for _ in range(10)]
print(f"10 d6 rolls: {rolls}")
# The valid results are the ones between 1 and 6
print(f"All rolls are valid: {all(1 <= value <= 6 for value in rolls)}")
```

Both `roll()` and `__call__()` managed to generate random dice values, where all 10 rolls from the six-sided dice were valid.

## Task 3: Weapon class

A `Weapon` object has a name and a `Dice` object used to calculate its damage. The `roll_damage()` method rolls the weapon's damage dice and returns a random damage value. This demonstrates composition: a weapon **has a** dice object.

- **Attributes:** `name`, `damage_dice`

- **Methods:** `roll_damage()`


```python
# A demonstration of the Weapon class

# Creates 2 weapons with different damage dice
sword = Weapon(name="Sword", damage_dice=Dice(sides=7))
axe = Weapon(name="Axe", damage_dice=Dice(sides=10))

print(f"{sword.name}: {sword.roll_damage()} damage")
print(f"{axe.name}: {axe.roll_damage()} damage")

# Compares armed and unarmed characters
armed = Character(name="Aria", weapon=sword)
unarmed = Character(name="Kura", damage=100)

# Creates the 2 targets
target1 = Character(name="Target 1")
target2 = Character(name="Target 2")
# Attacks the first target using weapon damage
armed.attack(target1)
# Attacks the second target using base damage
unarmed.attack(target2)

print(f"HP after weapon attack: {target1.hp}")
print(f"HP after base attack: {target2.hp}")
```

The sword and the axe initally dealt 6 and 9 damage respectively. Aria later dealt 1 damage with the sword, while Kura dealt their fixed base damage of 100. The results shows us how weapons use random damage, while the unarmed characters only use their base damage. 

## Task 4: Warrior and Mage subclasses

`Warrior` and `Mage` are specialised subclasses of the `Character` class. Both inherit attributes and methods from the parent class and use `super().__init__()` to reuse inherited attributes. They showcase method overriding through their combat abilities. `Warrior.take_damage()` reduces incoming damage using `armor` before calling `super().take_damage()`, while `Mage.attack()` uses `mana` to perform stronger magical attacks.

**Warrior:**

- **Additional attribute:** `armor`

- **Overridden method:** `take_damage()`

**Mage:**

- **Additional attribute:** `mana`, `mana_max`

- **Overridden method:** `attack()`


```python
# A demonstration of the subclasses

# Warrior
warrior = Warrior(name="Aria", hp_max=100, armor=5)
# Applies damage after armor reduction
warrior.take_damage(20)
print(f"Aria's HP after 20 incoming damage: {warrior.hp}")

# Mage
mage = Mage(name="Kura", hp_max=100, damage=10, mana=30)
target = Character(name="Target", hp_max=150)

# Tests out 5 mage attacks
for turn in range(1, 6):
    # Attacks and update the mage's mana
    mage.attack(target)
    print(f"Turn {turn}: Target's HP = {target.hp}, Kura's mana = {mage.mana}")
```

Aria's armor reduced the incoming damage of 20 to only 15, leaving them with 85 HP. Kura used mana for 3 stronger attacks before switching to the normal ones. These attacks also restored their mana.

## Task 5: Battle class

The `Battle` class manages turn-based combat between 2 `Character` objects, a `hero` and an `enemy`. The `play_round()` method allows the hero to attack first, followed by the enemy if they are still alive. `simulate_battle()` repeats rounds until one of the characters is defeated.

- **Attributes:** `hero`, `enemy`

- **Methods:** `play_round()`, `simulate_battle()` 


```python
# A demonstration of a single battle round
hero = Character(name="Aria", hp_max=100, damage=10)
enemy = Character(name="Kura", hp_max=100, damage=10)
# Sets up the battle
battle = Battle(hero=hero, enemy=enemy)

battle.play_round()
print(f"After 1 round: {hero.name} = {hero.hp} HP, {enemy.name} = {enemy.hp} HP")
```

As spesified, the code only generated 1 single battle round, where both characters had 90 HP remaining. It showcases how the hero attacked first, and the enemy following shortly.

## Battle simulations

The following demonstrations compare different character types, both with and without weapons. New characters are created for each scenario since HP and mana change during combat.

### Battle unarmed: Warrior vs Mage

```python
hero = Warrior(name="Aria", hp_max=100, damage=10, armor=5)
enemy = Mage(name="Kura", hp_max=100, damage=10, mana=5)

battle = Battle(hero=hero, enemy=enemy)
# Runs rounds until one dies
battle.simulate_battle()

# Finds the winner
winner = hero.name if hero.is_alive else enemy.name
print(f"Winner: {winner}")
print(f"Final HP: {hero.name} = {hero.hp}, {enemy.name} = {enemy.hp}")
print(f"Remaining mana: {enemy.mana}")
```

### Battle armed: Warrior vs Mage 

```python
# Uses the same random rolls each time
random.seed(123)

sword = Weapon(name="Sword", damage_dice=Dice(sides=10))
wand = Weapon(name="Wand", damage_dice=Dice(sides=5))

# The hero and the enemy each equipped with weapons
hero = Warrior(name="Aria", hp_max=100, damage=10, armor=5, weapon=sword)
enemy = Mage(name="Kura", hp_max=100, damage=10, mana=5, weapon=wand)

battle = Battle(hero=hero, enemy=enemy)
battle.simulate_battle()

winner = hero.name if hero.is_alive else enemy.name
print(f"Winner: {winner}")
print(f"Final HP: {hero.name} = {hero.hp}, {enemy.name} = {enemy.hp}")
print(f"Remaining mana: {enemy.mana}")
```

### Multiple battle simulations

This code runs several battles with new objects to show how random weapon damage can affect the results.

```python
# Uses a fixed seed for repeatable results
random.seed(2026)
# Starts the win counters at 0
results = {"Aria": 0, "Kura": 0}

# Runs 20 separate battles
for _ in range(20):
    # Creates a new sword, wand, hero and enemy for each battle
    sword = Weapon(name="Sword", damage_dice=Dice(sides=12)) 
    wand = Weapon(name="Wand", damage_dice=Dice(sides=8))
    hero = Warrior(name="Aria", hp_max=100, damage=10, armor=5, weapon=sword)
    enemy = Mage(name="Kura", hp_max=100, damage=10, mana=5, weapon=wand)
    
    # Sets up the next battle
    battle = Battle(hero=hero, enemy=enemy)
    battle.simulate_battle()

    winner = hero.name if hero.is_alive else enemy.name
    # Adds 1 win to the winner total
    results[winner] += 1

print("Winners of 20 battles:")
# Shows the wins for each character
for name, wins in results.items():
    print(f"{name}: {wins}")
```

In the unarmed battle, Aria won with 25 HP remaining. When both of the characters were armed, Kura won with 13 HP remaining. However, in the 20 continuous battles, Kura only won once.
